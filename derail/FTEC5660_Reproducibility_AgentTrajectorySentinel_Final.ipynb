```python
from pathlib import Path
import shutil

root = Path("/content/agent-trajectory-sentinel")
baseline = root / "student_results" / "baseline"
baseline.mkdir(parents=True, exist_ok=True)

files_to_copy = [
    (root / "results" / "tables" / "multiseed_summary.csv",
     baseline / "multiseed_summary.csv"),
    (root / "results" / "multiseed.json",
     baseline / "multiseed.json"),
    (root / "results" / "results.json",
     baseline / "master_seed_results.json"),
    (root / "baseline_full_log.txt",
     baseline / "baseline_full_log.txt"),
]

for source, destination in files_to_copy:
    shutil.copy2(source, destination)

for seed in [7, 101, 202, 303]:
    source = root / "results" / f"seed{seed}" / "results.json"
    destination = baseline / f"seed{seed}_results.json"
    shutil.copy2(source, destination)

print("Baseline saved successfully")
print("\nSaved files:")
for file in sorted(baseline.iterdir()):
    print("-", file.name)
```

```python
from pathlib import Path

file_path = Path("/content/agent-trajectory-sentinel/derail/experiments/run_experiment.py")
text = file_path.read_text(encoding="utf-8")

old_line = "FA_BUDGET = 0.05        # H1: healthy-val false-alarm budget for theta"
new_line = "FA_BUDGET = 0.10        # Student modification: relaxed H1 false-alarm budget"

if old_line not in text:
    raise ValueError("Original FA_BUDGET line was not found. Stop and check the file.")

print("Before:")
print(old_line)

text = text.replace(old_line, new_line, 1)
file_path.write_text(text, encoding="utf-8")

print("\nAfter:")
print(new_line)
print("\nModification completed successfully")
```
